## Semana 4 · Segmentar clientes y proponer acciones

**Objetivo:** describir perfiles de cliente que ayuden a decidir campañas, servicios o prioridades comerciales.

**Materiales:** notebook 03 · Segmentación; clientes y su definición en el diccionario.

**Tareas**

- [ ] Revisad nulos, valores extremos, recencia, frecuencia, gasto y devoluciones. Excluid el identificador y la fecha de la fotografía como variables de clustering.
- [ ] Justificad las variables elegidas. El gasto anual es ticket medio × pedidos: usar las tres puede dar demasiado peso al mismo comportamiento.
- [ ] Ejecutad las reglas del notebook como primera referencia. El notebook incluye una segmentación heurística; no contiene un clustering completo resuelto.

## Semana 5 · Hacer la prueba final y explicar el resultado

**Objetivo:** comprobar el modelo con datos reservados y comunicar sus errores y límites.

**Materiales:** notebook 02, apartados 5–6; entrada de julio; comparativa y decisiones congeladas en el CP3.

**Tareas**

- [ ] Dejad por escrito el modelo, los predictores y los parámetros antes de ejecutar la prueba.
- [ ] Evaluad enero–junio de 2026 una sola vez. No cambiéis el modelo para mejorar esta métrica y seguir presentándola como prueba independiente.
- [ ] Comparad el resultado con la validación y analizad al menos dos casos de error. Considerad el stock insuficiente, las diferencias de grupo y la información no observada.
- [ ] Explicad los factores que influyen en el modelo y los límites de la interpretación. Con datos sintéticos, una relación aprendida no demuestra una causa en el mercado real.

- [ ] Redactad una recomendación de stock y otra de campaña, con el grado de confianza que permitan las evidencias.
